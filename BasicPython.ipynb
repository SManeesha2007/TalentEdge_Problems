```python
n = int(input("Enter a number: "))

digits = []

while n > 0:
    digit = n % 10
    digits.append(digit)
    n = n // 10

# Sort digits in descending order
digits.sort(reverse=True)

# Form the result without using strings
result = 0

for digit in digits:
    result = result * 10 + digit

print(result)
```

```python
# printing reverse of a number in descending order of it's digits

n = int(input())

freq = [0]*10
temp = n

while temp>0:
    digit = temp%10
    freq[digit] += 1
    temp //=10

result = 0
for i in range(9, -1, -1):
    while freq[i]>0:
        result = result*10 + i
        freq[i] -= 1

print(result)
```

```python
def max_min(arr):
  maxi=arr[0]
  mini=arr[0]
  max_i=0;min_i=0
  for i in range(1,len(arr)):
    if maxi<arr[i]:
      maxi=arr[i]
      max_i=i
    if mini>arr[i]:
      mini=arr[i]
      min_i=i
  return maxi,max_i,mini,min_i

n=int(input())
arr=list(map(int,input().split()))
max,x,min,y=max_min(arr)
print(f"maximum number:{max} at index {x} ")
print(f"minimum number:{min} at index {y} ")
```

```python
def search(arr,key):
  index=-1
  for i in range(len(arr)):
    if arr[i]==key:
      index=i
      return index;
  return index
n=int(input())
arr=list(map(int,input().split()))
key=int(input())
res=search(arr,key)
if res!=-1:
  print("found at ",res)
else:
  print("not found")
```

```python
def nonrec_sum(a,b):
  return a+b
def rec_sum(a,b):
  if b==0:
    return a
  return rec_sum(a,b-1)+1

a,b=map(int,input().split())
print("non recursion : ",nonrec_sum(a,b))
print("recursion: ",rec_sum(a,b))
```

```python
#factorial in recursion
def fact(n):
  if n==0 or n==1:
    return 1
  return n*fact(n-1)
n=int(input())
if n<0:
  n=-n
print(f"factorial of {n} is {fact(n)}")
```
